# Exploração textual dos dados originais do INCT SiMAI

Formato, estrutura e primeiras linhas de `data_inct_original.json`. Os gráficos estão em `exploracao_grafica_percentual_inct.ipynb` (percentual) e `exploracao_grafica_quantitativa_inct.ipynb` (quantitativa).

```python
import json
from pathlib import Path

import pandas as pd

CAMINHO = Path('..', '..') / 'dados' / 'originais' / 'data_inct_original.json'

with open(CAMINHO, encoding='utf-8') as f:
    raw = json.load(f)

df = pd.DataFrame([
    {'nivel_m': float(nivel.removesuffix('m').replace('_', '.')),
     'bairro': bairro, **valores}
    for nivel, bairros in raw.items()
    for bairro, valores in bairros.items()
])

print(df.shape)
```
